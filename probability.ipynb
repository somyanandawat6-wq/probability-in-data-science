# Probability & Bayes Theorem - Python Implementation

## 1. What is Probability?

**Probability** is the measure of how likely an event is to occur. It ranges between **0** (impossible) and **1** (certain).

**Formula:**

P(E)=Number of favorable outcomes / Total number of possible outcomes

**Example: Tossing a fair coin**

- Outcomes = {Heads, Tails}
- P(Heads) = 1/2
- P(Tails) = 1/2

## 2. Types of Events

| Term | Description | Example |
|------|-------------|---------|
| **Independent** | One event doesn't affect the other | Tossing 2 coins |
| **Dependent** | One event affects the outcome of the next | Drawing cards without replacement |
| **Mutually Exclusive** | Events can't happen together | Rolling a die: Getting a 2 and a 5 at once |

## 3. Conditional Probability

**Definition**: The probability of an event A occurring **given** that event B has already occurred.

P(A∣B)=P(A∩B)P(B)P(A|B) = \frac{P(A \cap B)}{P(B)}P(A∣B)=P(B)P(A∩B)​

**Notation:**

- **P(A|B)** → "Probability of A given B" (conditional probability).
- **P(A∩B)** → "Probability of A and B" (joint probability).
- **P(B)** → "Probability of B."


**Example: Drawing Cards**

- Deck: 52 cards
- What's the probability of drawing an Ace **given** that the card is a Spade?
- Total Spade cards = 13
- Ace of Spades = 1

P(Ace∣Spade)=113P(Ace|Spade) = \frac{1}{13}P(Ace∣Spade)=131​


## 4. Bayes' Theorem

**Bayes' Theorem** helps us **reverse** conditional probabilities: What's the probability that cause A occurred given effect B?

**"The probability of A given B equals the probability of B given A times the probability of A, divided by the probability of B."**

Breakdown:

- **P(A|B)** → "Probability of A given B" (*posterior*).
- **P(B|A)** → "Probability of B given A" (*likelihood*).
- **P(A)** → "Probability of A" (*prior*).
- **P(B)** → "Probability of B" (*marginal probability*).


**Example: Medical Diagnosis (Disease Testing)**

- A disease affects **1 in 1,000 people**. (Prior probability of the disease P(D) = 0.001)
- **Test sensitivity**: 99% (If a person **has** the disease, the test shows positive with probability P(Positive|Disease) = 0.99)
- **False positive rate**: 5% (If a person **does not have** the disease, the test can still show positive with probability P(Positive|No Disease) = 0.05)
- **Question**: If a patient tests positive, what is the probability that they **actually have** the disease?


📌 **Inference**: Despite testing positive, the person has **only ~2% chance** of actually having the disease. This shows the **importance of prior probabilities** and how false positives can dominate when disease prevalence is low.

## Summary Table

| Concept | Formula/Description |
|---------|---------------------|
| **Basic Probability** | Favorable / Total outcomes |
| **Conditional Probability** | P(A \| B) = P(A ∩ B) / P(B) <br> P(A ∩ B): "P of A intersection B" |
| **Bayes' Theorem** | P(A \| B) = (P(B \| A) × P(A)) / P(B) |
| **Independent Events** | P(A ∩ B) = P(A) × P(B) <br> P(A ∩ B): "P of A intersection B" |
| **Mutually Exclusive** | P(A ∪ B) = P(A) + P(B), if A ∩ B = 0 <br> *"P of A union B"* <br> Probability that **at least one** of the two events (A or B) happens |


```python
from fractions import Fraction

# Define sample space and outcomes
coin_outcomes = ['Heads', 'Tails']
total_outcomes = len(coin_outcomes)  # Total possible outcomes = 2

# Calculate probabilities dynamically
prob_heads = Fraction(coin_outcomes.count('Heads'), total_outcomes)
prob_tails = Fraction(coin_outcomes.count('Tails'), total_outcomes)

print("Sample Space:", coin_outcomes)
print("P(Heads) =", prob_heads)  # Output: P(Heads) = 1/2
print("P(Tails) =", prob_tails)   # Output: P(Tails) = 1/2
```

# Case Study: Medical Diagnosis (Healthcare Domain)

**Bayes' Theorem** helps us **reverse** conditional probabilities: What's the probability that cause A occurred given effect B?

**"The probability of A given B equals the probability of B given A times the probability of A, divided by the probability of B."**

Breakdown:

- **P(A|B)** → "Probability of A given B" (*posterior*).
- **P(B|A)** → "Probability of B given A" (*likelihood*).
- **P(A)** → "Probability of A" (*prior*).
- **P(B)** → "Probability of B" (*marginal probability*).

- A disease affects **1 in 1,000 people**. (Prior probability of the disease P(D) = 0.001)
- **Test sensitivity**: 99% (If a person **has** the disease, the test shows positive with probability P(Positive|Disease) = 0.99)
- **False positive rate**: 5% (If a person **does not have** the disease, the test can still show positive with probability P(Positive|No Disease) = 0.05)
- **Question**: If a patient tests positive, what is the probability that they **actually have** the disease?

```python
print("\n--- Bayes Theorem: Medical Diagnosis ---")
# Given:
p_disease = 0.001 # 1 in 1000 people
p_positive_given_disease = 0.99 # 99% positive given disease
p_positive_given_no_disease = 0.05 # 95 % negative given no disease, so 5 % positive
p_no_disease = 1 - p_disease # probability of no disease is 1-disease

# Bayes Theorem
numerator = p_positive_given_disease * p_disease
denominator = (p_positive_given_disease * p_disease) + (p_positive_given_no_disease * p_no_disease)
p_disease_given_positive = numerator / denominator

print(f"P(Disease | Positive Test) ≈ {p_disease_given_positive:.4f} (~{round(p_disease_given_positive*100, 2)}%)")
```

### 4. Coin Example: 3 tosses, exactly 2 heads

```python
print("\n--- 3 Coin Tosses: Exactly 2 Heads ---")
from itertools import product

outcomes = list(product(['H', 'T'], repeat=3))

outcomes
```

product:- The product function and list serve specific purposes to generate all possible 

outcomes of a probability experiment (e.g., tossing a coin 3 times).

list:- Converts the generator returned by product into a list for easy viewing/use.

```python
favorable = [out for out in outcomes if out.count('H') == 2]
print("Total Outcomes:", len(outcomes))
print("Favorable Outcomes (2 Heads):", favorable)
print("P(2 Heads) =", Fraction(len(favorable), len(outcomes)))
```

### 5. Dice Example: Sum = 7

Dice Example:
Roll two 6-sided dice. What’s the probability the sum is 7?

```python
print("\n--- Dice Roll: Sum = 7 ---")
dice_outcomes = list(product(range(1, 7), repeat=2))
fav_dice = [pair for pair in dice_outcomes if sum(pair) == 7]
print("Total Outcomes:", len(dice_outcomes))
print("Favorable Outcomes (sum=7):", fav_dice)
print("P(Sum = 7) =", Fraction(len(fav_dice), len(dice_outcomes)))
```
